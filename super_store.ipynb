```python
import pandas as pd

df = pd.read_csv("SampleSuperstore.csv")

df.head()
df.info()
```

```python
#Sales By Region

sales_by_region = df.groupby('Region').Sales.sum()
sales_by_region
```

```python
#Find missing values and handle them
df.isnull().sum()

#Drop rows with missing values
#df_cleaned = df.dropna()

#Fill missing values
#df['Quantity'] = df['Quantity'].fillna(0)

#df
```

```python
#Create a new column Profit Margin (%) and find top 10 profitable transactions.

df['Profit_Margin'] = (df['Profit']/df['Sales'])*100

top_profit_margin = df.sort_values(by='Profit_Margin', ascending=False).head(10)
top_profit_margin
```

```python
#

avg_profit = df.groupby('Region')['Profit'].mean().sort_values(ascending=False)
avg_profit

hg_avg_profit = avg_profit.max()
top_region = avg_profit.idxmax()
print(top_region, hg_avg_profit)
```

```python
#Products with highest sales but below-average profit.

product_analysis = df.groupby('Sub-Category')[['Sales', 'Profit']].sum()

low_profit_products = product_analysis[
    product_analysis['Profit'] < product_analysis['Profit'].mean()
].sort_values(by='Sales', ascending=False)

low_profit_products.head(10)
```

```python
#High sales but loss-making transactions

loss_making_trans = df [
    (df['Sales'] > df['Sales'].mean()) & (df['Profit'] > 0)
]
loss_making_trans
```

```python
#Most profitable Category + Segment combination

cat_seg_profit = df.groupby(['Category', 'Segment'])['Profit'].sum().sort_values(ascending=False)
cat_seg_profit
```

```python
#3 regions that the company improve

region_rec = df.groupby('Region')[['Sales', 'Profit']].sum()
region_rec['profit_margin'] = (region_rec['Profit']/ region_rec['Sales'])*100
region_rec.sort_values(by='profit_margin', ascending=False).head(3)
```
